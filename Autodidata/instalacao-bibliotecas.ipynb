```python
import subprocess
import sys
from pathlib import Path

# Diretório centralizado das bibliotecas
caminho = Path("D:/Support/Library/")

# Bibliotecas que serão instaladas
bibliotecas = [
    "pathlib"
]

# Cria o diretório principal, caso não exista
caminho.mkdir(parents=True, exist_ok=True)

# Percorre a lista de bibliotecas
for biblioteca in bibliotecas:

    # Cria uma pasta individual para cada biblioteca
    destino = caminho / biblioteca
    destino.mkdir(parents=True, exist_ok=True)

    print(f"\nInstalando {biblioteca}...")

    # Instala a biblioteca no diretório correspondente
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        biblioteca,
        "--target",
        str(destino),
        "--upgrade"
    ])

    print(f"{biblioteca} instalada em: {destino}")

print("\nProcesso finalizado!")
```
